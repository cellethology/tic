```python
from tic.wrappers import FeatureWrapper

fw = FeatureWrapper(
    centre_types=tumor_list,
)
ft_adata = fw.fit(example_region)

```

```python
import pandas as pd


pd.DataFrame(ft_adata.obsm['composition']).head()
```

```python
# concat : composition and center_gene
import pandas as pd


composition = pd.DataFrame(ft_adata.obsm['composition'])
center_gene = pd.DataFrame(ft_adata.obsm['centre_gene'])
```

```python
from tic.wrappers import PseudotimeWrapper

ptw = PseudotimeWrapper(
    rep_key='centre_gene',
)

pt_adata = ptw.fit(ft_adata)
```

```python
ptw.metrics
```

```python
from tic.annotation.api import annotate_emt_state
from utils.dataset.codex_upmc import UPMC_EPITHELIAL_GENES, UPMC_MESENCHYMAL_GENES

tumor_adata = annotate_emt_state(adata=pt_adata, epithelial_genes=UPMC_EPITHELIAL_GENES,mesenchymal_genes=UPMC_MESENCHYMAL_GENES, cluster_on='scores', normalize_scores=True, general_score=True)
print(tumor_adata)

# plot the epi_score and mes_score where label is EMT

import scanpy as sc
import matplotlib.pyplot as plt

sc.pl.scatter(tumor_adata, x='epi_score', y='mes_score', color='emt_label')
```
