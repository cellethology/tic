```python
from tic.data import load_xenium_dataset

# Load subset of Xenium FFPE human breast dataset with a mask
kidney_adata = load_xenium_dataset(
    name='xenium_kidney_cancer',
    normalize='size',
    include_mask=True
)
print(kidney_adata)
```

```python
from tic.preprocessing.gene_variability import analyze_gene_variability
from tic.constant import EMT_GENES
analyze_gene_variability(adata=kidney_adata, gene_list=list(EMT_GENES), plot=True,viz_styles=['histogram_rug','bar_rank','scatter_rank','violin_points'])
```
