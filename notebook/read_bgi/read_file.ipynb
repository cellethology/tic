## find EMT genes in BGI dataset

```python
from tic.constant import EPITHELIAL_GENES, MESENCHYMAL_GENES, EMT_TF

gene_categories = {
    "Epithelial": EPITHELIAL_GENES,
    "Mesenchymal": MESENCHYMAL_GENES,
    "EMT Transcription Factors": EMT_TF,
}
gene_categories
```

## check variability of BGI dataset

```python
from tic.metrics import compute_variability

B03425E1_variability = compute_variability(B03425E1_adata,metrics=['cv','nonzero_ratio'])
B03425E1_tumor_variability = compute_variability(B03425E1_tumor_adata,metrics=['cv','nonzero_ratio'])
C03628C1_variability = compute_variability(C03628C1_adata,metrics=['cv','nonzero_ratio'])
C03628C1_tumor_variability = compute_variability(C03628C1_tumor_adata,metrics=['cv','nonzero_ratio'])
B03425E3_variability = compute_variability(B03425E3_adata,metrics=['cv','nonzero_ratio'])
B03425E3_tumor_variability = compute_variability(B03425E3_tumor_adata,metrics=['cv','nonzero_ratio'])
```

```python
from tic.plotting import plot_variability

for metric in ['cv', 'nonzero_ratio']:
    plot_variability(
        variability={
            "B03425E1": B03425E1_variability,
            "B03425E1_tumor": B03425E1_tumor_variability,
            "C03628C1": C03628C1_variability,
            "C03628C1_tumor": C03628C1_tumor_variability,
            "B03425E3": B03425E3_variability,
            "B03425E3_tumor": B03425E3_tumor_variability,
        },
        metric=metric,
        gene_categories=gene_categories,
        title=f"{metric} across BGI datasets",
        annotate_emt=True,
    )
```

## TIC pipeline

```python
from tic.graph.utils import estimate_radius
from tic.wrappers import FeatureWrapper, PseudotimeWrapper

radius = estimate_radius(adata=B03425E3_adata)
fw = FeatureWrapper(
    recipe='tme_default',
    centre_types=B03425E3_tumor_types,
    graph_params=None,
    subgraph_params={'strategy': 'radius', 'radius': radius},
)

B03425E3_fea_adata = fw.fit(B03425E3_adata)

print(B03425E3_fea_adata)
```

```python
pw = PseudotimeWrapper(
    rep_key='composition'
)

B03425E3_pt_adata = pw.fit(B03425E3_fea_adata)

print(B03425E3_pt_adata)
```

```python
from tic.plotting import plot_monotonicity_metrics_bar
plot_monotonicity_metrics_bar(
    metrics=pw.metrics,
    epithelial_genes=EPITHELIAL_GENES,
    mesenchymal_genes=MESENCHYMAL_GENES,
    emt_tf_genes=EMT_TF,
)
```

```python
from tic.plotting import plot_biomarker_trends
plot_biomarker_trends(
    adata=B03425E3_pt_adata,
    selected_biomarkers=['ACTA2','EPCAM'],
    y_transform='normalize'
)
```
